Titanic Dataset

## Analysis Performed
- Data loading and inspection
- Missing value analysis
- Statistical summary
- Survival analysis
- Gender-based survival analysis
- Passenger class analysis
- Age distribution analysis
- Correlation heatmap visualization

## Key Insights
- Female passengers had higher survival rates than males.
- First-class passengers had better survival chances.
- Most passengers were between 20–40 years old.
- Passenger class and fare influenced survival probability.

5. CHECK MISSING VALUES

```
import pandas as pd

url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
df = pd.read_csv(url)

df.isnull().sum()
```

6. BASIC STATISTICS

```
df.describe()
```

7. SURVIVAL COUNT GRAPH

```
import seaborn as sns
import matplotlib.pyplot as plt

sns.countplot(x='Survived', data=df)
plt.title("Survival Count")
plt.show()
```

8. MALE VS FEMALE SURVIVAL

```
sns.countplot(x='Survived', hue='Sex', data=df)

plt.title("Survival Based on Gender")

plt.show()
```

9. AGE DISTRIBUTION

```
plt.figure(figsize=(8,5))

sns.histplot(df['Age'].dropna(), bins=30)

plt.title("Age Distribution")

plt.show()
```

10. PASSENGER CLASS ANALYSIS

```
sns.countplot(x='Pclass', hue='Survived', data=df)

plt.title("Passenger Class vs Survival")

plt.show()
```

11. CORRELATION HEATMAP

```
numeric_df = df.select_dtypes(include=['number'])

plt.figure(figsize=(10,6))

sns.heatmap(numeric_df.corr(), annot=True)

plt.title("Correlation Heatmap")

plt.show()
```

12. OBSERVATIONS



1. Females had higher survival rates.
2. First class passengers survived more.
3. Most passengers were between 20–40 years old.
4. Fare and passenger class affected survival chances.
